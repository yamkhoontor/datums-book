## Listing 2.1  Ellipsoid constants and radii of curvature

````python
import math

# Defining parameters: semi-major axis a (m) and reciprocal flattening 1/f
ELLIPSOIDS = {
    "WGS84": (6378137.0, 298.257223563),
    "GRS80": (6378137.0, 298.257222101),
    "Everest 1830 (Modified)": (6377304.063, 300.8017),
}


def ellipsoid_constants(a, inv_f):
    """Derived constants of an ellipsoid from a and 1/f."""
    f = 1.0 / inv_f
    b = a * (1.0 - f)
    e2 = 2.0 * f - f * f          # first eccentricity squared
    ep2 = e2 / (1.0 - e2)         # second eccentricity squared
    n = f / (2.0 - f)             # third flattening
    return {"a": a, "f": f, "b": b, "e2": e2, "ep2": ep2, "n": n}


def dms_to_deg(d, m, s):
    """Degrees, minutes, seconds to decimal degrees."""
    sign = -1.0 if d < 0 else 1.0
    return sign * (abs(d) + m / 60.0 + s / 3600.0)


def radii_of_curvature(lat_deg, a, inv_f):
    """Meridian radius M, prime vertical radius nu, mean radius R."""
    c = ellipsoid_constants(a, inv_f)
    phi = math.radians(lat_deg)
    w2 = 1.0 - c["e2"] * math.sin(phi) ** 2
    nu = a / math.sqrt(w2)
    M = a * (1.0 - c["e2"]) / w2 ** 1.5
    R = math.sqrt(M * nu)
    return M, nu, R


if __name__ == "__main__":
    a, inv_f = ELLIPSOIDS["WGS84"]
    c = ellipsoid_constants(a, inv_f)
    print(f"b   = {c['b']:.4f} m")
    print(f"e2  = {c['e2']:.12f}")
    print(f"ep2 = {c['ep2']:.12f}")
    print(f"n   = {c['n']:.12f}")

    lat = dms_to_deg(1, 22, 0)
    M, nu, R = radii_of_curvature(lat, a, inv_f)
    print(f"M   = {M:.3f} m")
    print(f"nu  = {nu:.3f} m")
    print(f"R   = {R:.3f} m")
````

## Try it

Change the values between the two dashed lines and run the cell again. Nothing here alters the listings above.

````python
# ---- Change these values, then run the cell ----------------------
ellipsoid = "WGS84"            # "WGS84", "GRS80" or "Everest 1830 (Modified)"
lat_d, lat_m, lat_s = 1, 22, 0.0     # latitude: degrees, minutes, seconds
# -------------------------------------------------------------------

a, inv_f = ELLIPSOIDS[ellipsoid]
lat = dms_to_deg(lat_d, lat_m, lat_s)
M, nu, R = radii_of_curvature(lat, a, inv_f)
one_second = math.radians(1.0 / 3600.0)

print(f"Ellipsoid: {ellipsoid}   latitude: {lat:.9f} degrees")
print(f"M  = {M:.3f} m")
print(f"nu = {nu:.3f} m")
print(f"R  = {R:.3f} m")
print(f"one second of latitude  = {M * one_second:.3f} m")
print(f"one second of longitude = {nu * math.cos(math.radians(lat)) * one_second:.3f} m")
````
